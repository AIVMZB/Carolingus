```python
import cv2
import os
import matplotlib.pyplot as plt
```

```python
DATASET_ROOT = r"E:\Dyploma\Carolingus\Carolingus\datasets\words-in-lines-rotated-black\train"
```

```python
width_data = []
height_data = []
ratio_data = []

for file in os.listdir(DATASET_ROOT):
    if not file.endswith(".jpg"):
        continue
    
    file_path = os.path.join(DATASET_ROOT, file)
    line_image = cv2.imread(file_path)

    height, width = line_image.shape[:2]
    ratio = width / height

    width_data.append(width)
    height_data.append(height)
    ratio_data.append(ratio)


mean_width = sum(width_data) / len(width_data)
mean_height = sum(height_data) / len(height_data)
mean_ratio = sum(ratio_data) / len(ratio_data)

print(f"{mean_width=}")
print(f"{mean_height=}")
print(f"{mean_ratio=}")
```

```python
plt.hist(width_data, bins=30)
plt.grid()
plt.plot()
```

```python
plt.hist(height_data, bins=30)
plt.grid()
plt.plot()
```

```python
plt.hist(ratio_data, bins=30)
plt.grid()
plt.plot()
```
